```python
from abc import ABC, abstractmethod

class Operacao(ABC):
    def __init__(self, primeiro, segundo):
        self.primeiro = float(primeiro)
        self.segundo = float(segundo)
    @abstractmethod
    def executar(self):
        raise NotImplementedError

class Soma(Operacao):
    def executar(self):
        return self.primeiro + self.segundo

class Subtracao(Operacao):
    def executar(self):
        return self.primeiro - self.segundo

class Multiplicacao(Operacao):
    def executar(self):
        return self.primeiro * self.segundo

class Divisao(Operacao):
    def executar(self):
        if self.segundo == 0:
            raise ValueError("Não é possível dividir por zero.")
        return self.primeiro / self.segundo


# 3. Instanciando a classe filha e testando
conta = Soma(10, 5)
print(conta.executar())  # Saída: 15.0
```

```python
class Historico:
    def __init__(self):
        self.registros = []

    def adicionar(self, texto):
        self.registros.append(texto)

class Calculadora:
    def __init__(self):
        # Composição: a Calculadora cria e guarda um objeto Historico dentro dela
        self.historico = Historico()

    def calcular(self, Operacao):
        resultado = Operacao.executar()  # Polimorfismo acontecendo aqui!
        self.historico.adicionar(f"Resultado: {resultado}")
        return resultado

# Juntando tudo na prática:
calc = Calculadora()
print(calc.calcular(Soma(8, 2)))       # Saída: 10
print(calc.calcular(Subtracao(8, 2)))  # Saída: 6
print(calc.historico.registros)        # Saída: ['Resultado: 10', 'Resultado: 6']
```
